```python
import pandas as pd
import random

data = []

def generate_employee(performance):
    if performance == "Low":
        skill = random.randint(1, 2)
        dept_rating = random.randint(1, 2)
        attendance = random.randint(60, 75)
        absent = random.randint(10, 25)
        delay = random.randint(8, 20)
        experience = random.randint(0, 2)
        projects = random.randint(1, 5)
        feedback = random.randint(1, 2)
        training = random.randint(0, 10)
        overtime = random.randint(0, 5)
        errors = random.randint(8, 15)
        teamwork = random.randint(1, 2)

    elif performance == "Medium":
        skill = random.randint(3, 4)
        dept_rating = random.randint(3, 4)
        attendance = random.randint(76, 90)
        absent = random.randint(4, 9)
        delay = random.randint(3, 7)
        experience = random.randint(2, 6)
        projects = random.randint(6, 12)
        feedback = random.randint(3, 4)
        training = random.randint(10, 30)
        overtime = random.randint(5, 15)
        errors = random.randint(3, 7)
        teamwork = random.randint(3, 4)

    else:  # High
        skill = random.randint(4, 5)
        dept_rating = random.randint(4, 5)
        attendance = random.randint(91, 100)
        absent = random.randint(0, 3)
        delay = random.randint(0, 2)
        experience = random.randint(6, 15)
        projects = random.randint(13, 25)
        feedback = random.randint(4, 5)
        training = random.randint(30, 80)
        overtime = random.randint(15, 40)
        errors = random.randint(0, 2)
        teamwork = random.randint(4, 5)

    avg_task_time = round(random.uniform(4, 9), 2)

    productivity = (
        skill * 10 +
        dept_rating * 8 +
        feedback * 7 +
        attendance * 0.3 +
        projects * 2 -
        delay * 2 -
        errors * 3
    )

    return [
        delay,
        avg_task_time,
        dept_rating,
        attendance,
        absent,
        skill,
        experience,
        projects,
        feedback,
        training,
        overtime,
        errors,
        teamwork,
        round(productivity, 2),
        performance
    ]


# Generate 2000 rows per category
for perf in ["Low", "Medium", "High"]:
    for _ in range(2000):
        data.append(generate_employee(perf))


columns = [
    "task_delay_days",
    "avg_task_completion_time",
    "department_rating",
    "attendance_percentage",
    "days_absent",
    "skill_level",
    "years_experience",
    "projects_completed",
    "client_feedback_score",
    "training_hours",
    "overtime_hours",
    "errors_reported",
    "teamwork_score",
    "productivity_score",
    "performance"
]

df = pd.DataFrame(data, columns=columns)

# Save to CSV
df.to_csv("employee_performance_dataset.csv", index=False)

print("✅ Dataset generated successfully!")
print(df["performance"].value_counts())
```
